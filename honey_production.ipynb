```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("dataset.csv")
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
prod_per_year = df.groupby('year').totalprod.mean().reset_index()
```

```python
print(prod_per_year)
```

```python
X = prod_per_year['year']
X = X.values.reshape(-1, 1)
```

```python
Y = prod_per_year['totalprod']
```

```python
plt.scatter(X,Y)
plt.show()
```

```python
regression_model = linear_model.LinearRegression()
```

```python
regression_model.fit(X, Y)
```

```python
pip install joblib
```

```python
from joblib import dump
import os

model_dir = 'model'
os.makedirs(model_dir, exist_ok=True)

model_path = os.path.join(model_dir, 'honey_model.pkl')

dump(regression_model, model_path)

print(f"Model successfulyy saved to :{model_path} ")
print(f"File exists: {os.path.exists(model_path)}")
```

```python
print(regression_model.coef_)
```

```python
print(regression_model.intercept_)
```

```python
y_predict = regression_model.predict(X)
```

```python
plt.scatter(X, Y, color='blue', label='Actual data')
plt.plot(X, y_predict, color='red', label='Regression line')

plt.xlabel('Year')
plt.ylabel('Production per year')
plt.title('Actual Data vs Predicted Production')
plt.legend()
plt.show()
```

```python
X_future = np.array(range(2013, 2050))
X_future = X_future.reshape(-1, 1)
```

```python
X_predict = regression_model.predict(X_future)
```

```python
plt.plot(X, y_predict)
plt.plot(X_future, X_predict, color='red')
plt.xlabel('Year')
plt.ylabel('Production per year')
plt.title('Predicted Production')

plt.show()
```
